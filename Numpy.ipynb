# **Airtmetic Operations**

# **Slicing**

```python
# 1D Array
var = np.array([1,2,3,4,5,6,7])
print(var)
print('2 to 5: ',var[1:5])
print('2 to End: ',var[1:])
print('1 to 5: ',var[:5])
print('step: ',var[::2])
print('step: ',var[1:4:2])
```

# **Iterating Arrays**

```python
# 1D  Array
var=np.array([9,8,7,6,5,4,3])
print(var)
for i in var:
  print(i)
```

```python
# 2D Array
var = np.array([[1,2,3,4],[9,8,7,6]])
print(var,'\n')
for j in var:
  print(j)
print()
for i in var:
  for j in i:
    print(j)
```

```python
# 3D Array
var = np.array([[[3,2,1],[6,5,4]]])
print(var)
print()
# for i in var:
#   for j in i:
#     for k in j:
#       print(k)
for i in np.nditer(var):
  print(i)
print()
for i in np.ndenumerate(var):
  print(i)
```

# **Copy and View**

```python
var = np.array([1,2,3,4])
co=var.copy()
print('var: ',var)
print('copy: ',co)
co[1]=40
print()
print('var: ',var)
print('copy: ',co)
```

```python
var = np.array([1,2,3,4])
vi=var.view()
print('var: ',var)
print('View: ',vi)
vi[1]=40
print()
print('var: ',var)
print('View: ',vi)
```

# **Join Array**

```python
# 1D Array
var = np.array([1,2,3,4])
var2= np.array([5,6,7,8])
ar= np.concatenate((var,var2))
print('var: ',var)
print('var2: ',var2)
print('ar: ',ar)
```

```python
# 2D Array
var = np.array([[1,2],[3,4]])
var2= np.array([[5,6],[7,8]])
ar= np.concatenate((var,var2),axis=1)
print('var: \n',var)
print('var2: \n',var2)
print('ar: \n',ar)
```

# **Matrix Function**

```python
# Transpose
var = np.matrix([[1,2,3],[4,5,6]])
print(var)
print()

print(np.transpose(var))
```

```python
# Swapaxes
var = np.matrix([[1,2,3],[4,5,6]])
print(var)
print()

print(np.swapaxes(var,0,1))
```

```python
# Inverse
var = np.matrix([[1,2],[3,4]])
print(var)
print()

print(np.linalg.inv(var))
```

```python
# Power of Matrix
var = np.matrix([[1,2],[3,4]])
print(var)
print()
print(np.linalg.matrix_power(var,2))
print()
print(np.linalg.matrix_power(var,0))
print()
print(np.linalg.matrix_power(var,-2))
```

```python
# Determinant
var = np.matrix([[1,2],[3,4]])
print(var)
print()
print(np.linalg.det(var))
print()
var = np.matrix([[1,2,3],[3,4,3],[4,5,6]])
print(var)
print()
print(np.linalg.det(var))
```
