# **Series**

```python
import pandas as pd
```

```python
x=[3,4,5,6,7,8]
var = pd.Series(x)
print(var)
print(type(var))
print(var[2])
```

```python
x=[3,4,5,6,7,8]
var = pd.Series(x,index=['a','b','c','d','e','f'])
print(var)
print(type(var))
print(var[2])
```

```python
x=[3,4,5,6,7,8]
var = pd.Series(x,index=['a','b','c','d','e','f'],dtype='float')
print(var)
print(type(var))
print(var[2])
```

```python
x=[3,4,5,6,7,8]
var = pd.Series(x,index=['a','b','c','d','e','f'],dtype='float',name='Python')
print(var)
print(type(var))
print(var[2])
```

```python
dic={'name':['Python','C++','C','Java'],'popularity':[12,13,14,15],'Rank':[1,4,3,2]}
var= pd.Series(dic)
print(var)
```

```python
s=pd.Series(12)
print(s)
print(type(s))
```

```python
s=pd.Series(12,index=[1,2,3,4,5,6])
print(s)
print(type(s))
```

```python
s1=pd.Series(12,index=[1,2,3,4,5,6])
s2=pd.Series(12,index=[1,2,3,4])
print(s1+s2)
```

# **DataFrame**

```python
l=[1,2,3,4,5,6]
var=pd.DataFrame(l)
print(var)
print(type(var))
```

```python
dic={'a':[1,2,3,4,5],'s':[1,2,3,4,5]}
var=pd.DataFrame(dic)
print(var)
print(type(var))
```

```python
dic={'a':[1,2,3,4,5],'s':[1,2,3,4,5]}
var=pd.DataFrame(dic,columns=['a'])
print(var)
print(type(var))
```

```python
dic={'a':[1,2,3,4,5],'s':[1,2,3,4,5],'d':[1,2,3,4,5],1:[1,2,3,4,5]}
var=pd.DataFrame(dic,columns=['a',1])
print(var)
print(type(var))
```

```python
dic={'a':[1,2,3,4,5],'s':[1,2,3,4,5],'d':[1,2,3,4,5],1:[1,2,3,4,5]}
var=pd.DataFrame(dic,columns=['a',1],index=['a','s','d','g','f'])
print(var)
print(type(var))
```

```python
dic={'a':[1,2,3,4,5],'s':[1,2,3,4,5],'d':[1,2,3,4,5],1:[1,2,3,4,5]}
var=pd.DataFrame(dic)
print(var)
print(var['a'][3])
print(type(var))
```

```python
list=[[1,2,3,4,5],[11,12,13,14,15]]
var =pd.DataFrame(list)
print(var)
print(type(var))
```

```python
sr={'s':pd.Series([1,2,3,4]),'r':pd.Series([1,2,3,4])}
var=pd.DataFrame(sr)
print(var)
print(type(var))
```

# Arithmetic **operations**

```python
var=pd.DataFrame({'A':[1,2,3,4],'B':[5,6,7,8]})
print(var)
```

```python
var['C']=var['A']+var['B']
var
```

```python
var['C']=var['A']-var['B']
var
```

```python
var['C']=var['A']*var['B']
var
```

```python
var['C']=var['A']/var['B']
var
```

```python
var=pd.DataFrame({'A':[10,20,30,40],'B':[15,16,17,18]})
print(var)
```

```python
var['(<=20)']=var['A']<=20
print(var)
```

# **Insert**

```python
var=pd.DataFrame({'A':[1,2,3,4,5],'C':[9,8,7,6,5]})
print(var)
```

```python
var.insert(1,'B',var['A'])
print(var)
```

```python
var.insert(3,'D',[11,12,13,14,15])
print(var)
```

```python
var['D']=var['A'][:3]
print(var)
```
